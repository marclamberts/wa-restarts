```python
# WA set pieces, restarts, xDanger and reports: point to a league folder of IMPECT match JSON, pick clubs, reports,
# style and credits, run this cell.
LEAGUE_FOLDER = "events_json/3. Liga 2026-2027"   # one league-season's match JSON
CLUBS = []                     # full or partial club names, e.g. ["Ajax"]; [] = league only

# reports (PDF + one PNG per page in OUT/<league>/reports/)
REPORTS = ["post_match", "player", "set_piece", "restart"]   # any subset; [] = no reports
MATCH = "latest"                     # post-match + player match reports: "latest" (each club's last match), "Home vs Away" or a match id
PLAYERS = []                         # player reports, e.g. ["Gloukh", "Taylor"] (partial names work)
PLAYER_REPORTS = ["match", "season"] # match report for MATCH, season report, or both
SCOPE = "both"                       # set-piece + restart reports: "team" (each club), "league" or "both"
FORMS = ["short", "long"]            # post-match + player match reports: 1-page short form, multi-page long form

# look
STYLE = "navy"                       # navy, fog, dusk, cream, clay, espresso, xhotspur; a list writes the reports in each
FONT = "house"                       # house (Georgia + Helvetica Neue), plex, spectral, display, barlow; a list works too

# credits on every chart
AUTHOR = "Marc Lamberts"             # made by: header and footer
BRAND = "Waltzing Analytics"         # wordmark in the header
MARK = "WA"                          # short mark in the header
DATA = "IMPECT"                      # data credit in the source line
MODELS_BY = "Waltzing Analytics"     # model credit in the source line

CHARTS = True                        # league folder, club set-piece packs and xDanger tables (as before)
XDANGER = True                       # with CHARTS: also xDanger per action, player and team in OUT/<league>/xdanger/
REFRESH = False                      # True: rescore from the JSON even if cached (new match files are picked up anyway)
OUT = "wa_restarts_output"

from pathlib import Path
import wa_restarts as wr

styles = [STYLE] if isinstance(STYLE, str) else list(STYLE)
fonts = [FONT] if isinstance(FONT, str) else list(FONT)
folder = Path(LEAGUE_FOLDER)
LEAGUE = folder.name                 # the folder name is the league-season name
wr.configure(json_root=folder.parent, out_root=OUT,   # events_meta is looked for next to the JSON root
             style=styles[0], font=fonts[0], author=AUTHOR, brand=BRAND, mark=MARK, data=DATA, models_by=MODELS_BY)
print(f"{LEAGUE}: {len(list(folder.glob('*.json')))} matches  ·  styles {styles}  ·  fonts {fonts}")
print("Clubs:", wr.teams(LEAGUE))
try:
    if CHARTS:
        wr.build(LEAGUE, teams=CLUBS, refresh=REFRESH, xdanger_tables=XDANGER)
    if REPORTS:
        pdfs = wr.reports(LEAGUE, kinds=REPORTS, clubs=CLUBS, match=MATCH, players=PLAYERS, player_reports=PLAYER_REPORTS,
                          scope=SCOPE, forms=FORMS, styles=styles, fonts=fonts)
        print(f"{len(pdfs)} report PDFs")
except (ValueError, FileNotFoundError) as e:   # unknown club, player or match, or an empty folder
    print(e)
```
