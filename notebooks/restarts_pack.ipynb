```python
# Set-piece and restart pack: point to a league folder of IMPECT match JSON, pick clubs, style and credits, run this cell.
# League folder (O_* charts, B01-B24 visual pack, CSVs) + one set-piece pack per club land in OUT/<league>/
LEAGUE_FOLDER = "events_json/3. Liga 2026-2027"   # one league-season's match JSON
CLUBS = []                           # full or partial club names, e.g. ["Essen"]; [] = league folder only
STYLE = "navy"                       # navy, fog, dusk, cream, clay, espresso, xhotspur
XDANGER = True                       # also write xDanger per action, player and team to OUT/<league>/xdanger/

# credits on every chart
AUTHOR = "Marc Lamberts"             # made by: header and footer
BRAND = "Waltzing Analytics"         # wordmark in the header
MARK = "WA"                          # short mark in the header
DATA = "IMPECT"                      # data credit in the source line
MODELS_BY = "Waltzing Analytics"     # model credit in the source line

REFRESH = False                      # True: rescore from the JSON even if cached (new match files are picked up anyway)
OUT = "wa_restarts_output"

from pathlib import Path
import wa_restarts as wr

folder = Path(LEAGUE_FOLDER)
LEAGUE = folder.name                 # the folder name is the league-season name
wr.configure(json_root=folder.parent, out_root=OUT,   # events_meta is looked for next to the JSON root
             style=STYLE, author=AUTHOR, brand=BRAND, mark=MARK, data=DATA, models_by=MODELS_BY)
print(f"{LEAGUE}: {len(list(folder.glob('*.json')))} matches  ·  style {STYLE}")
print("Clubs:", wr.teams(LEAGUE))
try:
    folders = wr.build(LEAGUE, teams=CLUBS, refresh=REFRESH, xdanger_tables=XDANGER)
except (ValueError, FileNotFoundError) as e:   # unknown club or empty folder
    print(e)
```
